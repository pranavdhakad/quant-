```python
import yfinance as yf
import pandas as pd

# 1. Pull 3 years of daily historical data for HDFC Bank on NSE
ticker = "HDFCBANK.NS"
data = yf.download(ticker, start="2023-11-10", end="2026-11-10")

# 2. Isolate the 'Close' price and calculate day-over-day percentage returns
data['Daily_Return'] = data['Close'].pct_change()

# 3. Clean up the missing data row and display the terminal view
clean_data = data[['Close', 'Daily_Return']].dropna()
print(f"--- Quant Dataset for {ticker} ---")
print(clean_data.tail(5))
```
